```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import scipy
from scipy.optimize import curve_fit
```

```python
d1 = [28,70,109,150,190]
d2 = [29,69,110,150,190]
d3 = [30,70,110,150,190]
N = [1,2,3,4,5]
d = []
std = []
for x in range(len(d3)):
    #print(x)
     d.append(np.mean([d1[x],d2[x],d3[x]]))
     std.append(np.std([d1[x],d2[x],d3[x]]))
df = pd.DataFrame()
df["markings"] = d
df["cycles"] = N
df["standard deviation distances"] = std

df["actual distance"] = df["markings"]/df["cycles"]
```

```python
df["micrometers"] = df["actual distance"].apply(lambda a: a*2)
```

```python
df["D_nano"] = df["micrometers"].apply(lambda b: b*1000)
```
