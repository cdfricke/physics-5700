```python
import glob
import pandas as pd
from analysis import analysis
import sys
import numpy as np

def a_err(vg):
    eta = 18.13e-6 #N*s/m^2
    g = 9.801 #m/s^2
    rho = 0.86e3 #kg/m^3
    return 1/2 * np.sqrt((9*eta)/(2*g*rho*vg))
```

```python
analyzer = analysis()
analyzer.load("drop_data.csv")
```

```python
df = pd.read_csv("drop_data.csv")

df["A_err"] = df["VG"].apply(a_err)
```
