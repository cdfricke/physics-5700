```python
import numpy as np

d1 = np.array([28, 70, 109, 150, 190])
d2 = np.array([29, 69, 110, 150, 190])
d3 = np.array([30, 70, 110, 150, 190])

d1 = (d1 - d1[0]) * 2.0
d2 = (d2 - d2[0]) * 2.0
d3 = (d3 - d3[0]) * 2.0
```

```python
for i in range(1, len(d1)):
    d1[i] /= i
    d2[i] /= i
    d3[i] /= i

d1 = d1[1:]
d2 = d2[1:]
d3 = d3[1:]
```

```python
measurements = np.concat((d1, d2, d3))
print(measurements)
```

```python
mean = np.mean(measurements)
std = np.std(measurements)
total_err = np.sqrt(std**2 + 1.0**2)
print("d =", mean, "+-", total_err, "um.")3
```

```python
lambda_avg = np.mean([576.9598, 579.0663]) / 1e3
expected_val = np.abs(576.9598 - 579.0663)
print(expected_val)
```

```python
def dL(d, lAvg):
    return lAvg*lAvg / (2*d)

def dLdD(d, lAvg):
    return lAvg*lAvg / (2*d*d)

result = dL(mean, lambda_avg)
result_err = std*(dLdD(mean, lambda_avg))
print("deltaL =", result*1e3, "+-", result_err*1e3)
```
