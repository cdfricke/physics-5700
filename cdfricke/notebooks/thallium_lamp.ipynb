## Analysis for Measuring Wavelength of the Thallium Green Lamp!
---
The expected value is $\lambda = 535 \ \rm nm$.

```python
import numpy as np
import matplotlib.pyplot as plt

N = np.array([63, 63, 69, 70, 70, 68, 72, 69, 60, 69, 72, 65, 73, 71])
N_avg = np.mean(N)
N_err = np.std(N)

d_avg = 20.0 # microns
d_err = 2.0
```

```python
fig, ax = plt.subplots()
fig = plt.hist(N)
plt.axvline(x=N_avg, linestyle='dashed', color='black')
plt.show()
```

```python
def WL(d, N) -> float:
    return 2.0 * d / N

def dWL_dD(N) -> float:
    return 2.0 / N

def dWL_dN(d, N) -> float:
    return -2.0 * d / (N**2)
```

```python
wl_avg = WL(d_avg, N_avg)

# error propagation
term1 = d_err**2 * dWL_dD(N_avg)**2
term2 = N_err**2 * dWL_dN(d_avg, N_avg)**2
wl_err = np.sqrt(term1 + term2)
```

```python
# thallium green lamp results
print('The wavelength of the lamp is', round(wl_avg*1e3), '+-', round(wl_err*1e3), 'nm.')
```

```python
# Now what happens if we assume we undercounted? Let's correct each count by a factor of 1.1
N = N * 1.1
N_avg = np.mean(N)
N_err = np.std(N)

fig, ax = plt.subplots()
fig = plt.hist(N)
plt.axvline(x=N_avg)
plt.show()
```

```python
wl_avg = WL(d_avg, N_avg)

# error propagation
term1 = d_err**2 * dWL_dD(N_avg)**2
term2 = N_err**2 * dWL_dN(d_avg, N_avg)**2
wl_err = np.sqrt(term1 + term2)
```

```python
# thallium green lamp results
print('The wavelength of the lamp is', round(wl_avg*1e3), '+-', round(wl_err*1e3), 'nm.')
```
